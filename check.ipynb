```python
import pandas as pd
import os

base_files_path = 'results/questions/logs_base'
base_files = os.listdir(base_files_path)
```

```python
list(set([x.split('_')[0] for x in base_files]))
```

```python
from collections import defaultdict
model2scores = defaultdict(list)
```

```python
import re

for split in range(10):
    for model in ['GT-sep', 'GAT-sep', 'GAT', 'TAG', 'GCN', 'SAGE', 'GT', 'ResNet']:
        files = [
            x for x in base_files
            if f'{model}_' in x and f'split{split}' in x
        ]
        if len(files) == 0:
            continue
        scores = {}
        for file in files:
            with open(f'{base_files_path}/{file}', 'r') as f:
                content = f.read()
                train_match = re.search(r'train_acc=([\d.]+)', content)
                val_match = re.search(r'val_acc=([\d.]+)', content)
                test_acc_match = re.search(r'test_acc=([\d.]+)', content)
                test_f1m_match = re.search(r'test_f1m=([\d.]+)', content)
                test_f1w_match = re.search(r'test_f1w=([\d.]+)', content)
                test_prec_match = re.search(r'test_prec=([\d.]+)', content)
                test_rec_match = re.search(r'test_rec=([\d.]+)', content)
                test_loss_match = re.search(r'test_loss=([\d.]+)', content)
                scores[file] = {
                    'train_acc': float(train_match.group(1)) if train_match else None,
                    'val_acc': float(val_match.group(1)) if val_match else None,
                    'test_acc': float(test_acc_match.group(1)) if test_acc_match else None,
                    'test_f1m': float(test_f1m_match.group(1)) if test_f1m_match else None,
                    'test_f1w': float(test_f1w_match.group(1)) if test_f1w_match else None,
                    'test_prec': float(test_prec_match.group(1)) if test_prec_match else None,
                    'test_rec': float(test_rec_match.group(1)) if test_rec_match else None,
                    'test_loss': float(test_loss_match.group(1)) if test_loss_match else None
                }
        best_val_acc = 0
        for file, score in scores.items():
            if score['val_acc'] > best_val_acc:
                best_val_acc = score['val_acc']
                best_file = file
        print(f'{model}\t {best_file}\t test acc: {scores[best_file]["test_acc"]}\t stats: {scores[best_file]}')
        model2scores[model + "_BASE"].append(scores[best_file]["test_acc"])
    print()
    print('=' * 100)
    print()
```

```python
import pandas as pd
import re
from pathlib import Path

# models = ['GT-sep', 'GAT-sep', 'GAT', 'TAG', 'GCN', 'SAGE', 'GT', 'ResNet']
models = ['GT-sep', 'GAT-sep', 'GAT', 'GCN', 'SAGE', 'GT', 'ResNet']

root = Path('.')
roman_ens_path = root / 'logs_ensemble'
if not roman_ens_path.exists():
    roman_ens_path = root / 'roman_empire' / 'logs_ensemble'

# For these datasets, aggregate test ROC-AUC (from the *selected* layer block), not test accuracy.
ROC_PRIMARY_DATASETS = frozenset({'questions', 'minesweeper', 'tolokers'})

dataset_log_dirs = {
    'roman_empire': {
        'BASE': root / 'results' / 'roman_empire' / 'logs_base',
        'ENS': root / 'results' / 'roman_empire' / 'logs_ensemble',
        'TABM': root / 'results' / 'roman_empire' / 'logs_tabm',
    },
    'amazon_ratings': {
        'BASE': root / 'results' / 'amazon_ratings' / 'logs_base',
        'ENS': root / 'results' / 'amazon_ratings' / 'logs_ensemble',
        'TABM': root / 'results' / 'amazon_ratings' / 'logs_tabm',
    },
    'minesweeper': {
        'BASE': root / 'results' / 'minesweeper' / 'logs_base',
        'ENS': root / 'results' / 'minesweeper' / 'logs_ensemble',
        'TABM': root / 'results' / 'minesweeper' / 'logs_tabm',
    },
    'questions': {
        'BASE': root / 'results' / 'questions' / 'logs_base',
        'ENS': root / 'results' / 'questions' / 'logs_ensemble',
        'TABM': root / 'results' / 'questions' / 'logs_tabm',
    },
    'tolokers': {
        'BASE': root / 'results' / 'tolokers' / 'logs_base',
        'ENS': root / 'results' / 'tolokers' / 'logs_ensemble',
        'TABM': root / 'results' / 'tolokers' / 'logs_tabm',
    },
}

def _layer_block_for(content, layer):
    """Body of one depth block: from `NL split=` header through the next depth header (or EOF)."""
    # NOTE: can't write ={60,} inside an f-string — it becomes the tuple (60,).
    bar = '=' * 60
    start_pat = re.compile(
        rf'(?ms)^{bar}\s*\n[^\n]*(?<!\d){layer}L split=[^\n]*\n{bar}\s*\n'
    )
    m = start_pat.search(content)
    if not m:
        return None
    rest = content[m.end() :]
    end_pat = re.compile(rf'(?ms)^{bar}\s*\n[^\n]*(?<!\d)\d+L split=')
    end_m = end_pat.search(rest)
    if end_m:
        return rest[: end_m.start()]
    return rest


def _test_roc_auc_from_final_block(content):
    """BASE logs: ROC-AUC is printed under `--- Final ... ---` near the end (before `>> best_step=`)."""
    marker = '--- Final'
    idx = content.rfind(marker)
    window = content[idx:] if idx != -1 else content
    rocs = re.findall(r'test_roc_auc=([\d.]+)', window)
    if rocs:
        return float(rocs[-1])
    rocs_all = re.findall(r'test_roc_auc=([\d.]+)', content)
    return float(rocs_all[-1]) if rocs_all else None


def _test_roc_auc_from_layer_block(block):
    """TABM/ENS: take ROC from that depth's summary tail (`--- Ensemble` or `--- Final`), not from training traces."""
    if not block:
        return None
    ei = block.rfind('--- Ensemble')
    fi = block.rfind('--- Final')
    idx = max(ei, fi)
    tail = block[idx:] if idx != -1 else block
    rocs = re.findall(r'test_roc_auc=([\d.]+)', tail)
    return float(rocs[-1]) if rocs else None


def parse_scores(log_file, primary_is_roc: bool = False):
    """
    For TABM/ENS: use the *last* `>> selected best layer=N ...` line in the file (EOF summary).
    `val_acc` on that line drives comparison across *different log files* for the same split;
    `test_roc_auc` is read only from that layer's `--- Ensemble` / `--- Final` block (never argmax ROC).

    primary_is_roc: for questions/minesweeper/tolokers, the reported metric is test_roc_auc
    from that selected layer block; otherwise we use test_acc.
    """
    content = Path(log_file).read_text()

    selected_re = re.compile(
        r'>>\s*selected best layer=(\d+)(?:\s+best_step=\d+)?\s+val_acc=([\d.]+)\s+test_acc=([\d.]+)'
    )
    sel_matches = selected_re.findall(content)
    if sel_matches:
        # Use the *last* `>> selected best layer=...` in the file (authoritative summary at EOF),
        # not argmax val_acc over matches — ROC must match that chosen layer, not a "best ROC" layer.
        layer_s, val_s, test_s = sel_matches[-1]
        layer = int(layer_s)
        val_acc = float(val_s)
        test_acc = float(test_s)
        block = _layer_block_for(content, layer)
        roc = _test_roc_auc_from_layer_block(block)
        if primary_is_roc and roc is not None:
            primary = roc
        else:
            primary = test_acc
        return {
            'val_acc': val_acc,
            'test_acc': test_acc,
            'test_roc_auc': roc,
            'primary': primary,
        }

    # BASE (single layer per file): best checkpoint summary at the end.
    best_ckpt_match = re.search(
        r'>>\s*best_step=\d+\s+val_acc=([\d.]+)\s+test_acc=([\d.]+)',
        content,
    )
    if best_ckpt_match:
        val_acc = float(best_ckpt_match.group(1))
        test_acc = float(best_ckpt_match.group(2))
        # For ROC-primary datasets, take test_roc_auc from the final summary block (end of log).
        roc = _test_roc_auc_from_final_block(content)
        if primary_is_roc and roc is not None:
            primary = roc
        else:
            primary = test_acc
        return {
            'val_acc': val_acc,
            'test_acc': test_acc,
            'test_roc_auc': roc,
            'primary': primary,
        }

    val_accs = re.findall(r'val_acc=([\d.]+)', content)
    test_accs = re.findall(r'test_acc=([\d.]+)', content)
    val_acc = float(val_accs[-1]) if val_accs else None
    test_acc = float(test_accs[-1]) if test_accs else None
    roc = _test_roc_auc_from_final_block(content)
    primary = None
    if val_acc is not None:
        if primary_is_roc and roc is not None:
            primary = roc
        else:
            primary = test_acc
    return {
        'val_acc': val_acc,
        'test_acc': test_acc,
        'test_roc_auc': roc,
        'primary': primary,
    }

rows = []
for dataset_name, method2dir in dataset_log_dirs.items():
    primary_is_roc = dataset_name in ROC_PRIMARY_DATASETS
    for method_name, logs_dir in method2dir.items():
        if not logs_dir.exists():
            continue

        all_files = [x.name for x in logs_dir.iterdir() if x.is_file() and x.suffix == '.log']

        for model in models:
            split_scores = []
            for split in range(10):
                files = [
                    x for x in all_files
                    if f'{model}_' in x and f'split{split}' in x
                ]


                if not files:
                    continue

                best_val = -1.0
                best_primary = None

                for file_name in files:
                    score = parse_scores(logs_dir / file_name, primary_is_roc=primary_is_roc)
                    val_acc = score['val_acc']
                    if val_acc is None:
                        continue
                    if val_acc > best_val:
                        best_val = val_acc
                        # best_primary = score['primary']
                        best_primary = score['test_acc']

                if best_primary is not None:
                    split_scores.append(best_primary)

            if split_scores:
                rows.append({
                    'model_variant': f'{model}_{method_name}',
                    'dataset': dataset_name,
                    'mean_acc': float(pd.Series(split_scores).mean()),
                    'std_acc': float(pd.Series(split_scores).std(ddof=0)),
                })

mx_mean = pd.DataFrame(rows).pivot_table(
    index='model_variant',
    columns='dataset',
    values='mean_acc',
    aggfunc='mean',
).sort_index().mul(100)

mx_std = pd.DataFrame(rows).pivot_table(
    index='model_variant',
    columns='dataset',
    values='std_acc',
    aggfunc='mean',
).reindex(mx_mean.index).mul(100)

mx_dataset = mx_mean.round(3)

mx_text = mx_mean.round(3).astype(str) + ' ± ' + mx_std.round(3).astype(str)

def highlight_top2(col):
    styles = ['' for _ in col]
    non_na = col.dropna()
    if non_na.empty:
        return styles

    unique_vals = sorted(non_na.unique(), reverse=True)
    best = unique_vals[0]
    second = unique_vals[1] if len(unique_vals) > 1 else None

    for i, v in enumerate(col):
        if pd.isna(v):
            continue
        if v == best:
            styles[i] = 'background-color: #ffd700; font-weight: 700'  # gold
        elif second is not None and v == second:
            styles[i] = 'background-color: #c0c0c0; font-weight: 700'  # silver
    return styles

mx_text.style.apply(lambda col: highlight_top2(mx_mean[col.name]), axis=0)
```
